```python
import cv2 as cv
import os.path

# Start a video capture, using device's camera
cap = cv.VideoCapture(0)

# Check if video file opened successfully
if (cap.isOpened() == False):
    print("Error opening video stream or file")
frame_width = int(cap.get(3))
frame_height = int(cap.get(4))

print("Frame width: " , frame_width)
print("Frame height: " , frame_height)
count = 0
# Read until video is completed
while(cap.isOpened()):
    # Capture frame-by-frame
    ret, frame = cap.read()
    if ret == False:
        break

    # Display the frame
    cv.imshow('frame', frame)
    key = cv.waitKey(25)

    if key & 0xFF == ord('x'):
        filename = f"image0{count}.jpg"
        while os.path.isfile(filename):
            count+=1
            filename = f"image0{count}.jpg"

        cv.imwrite(filename, frame)
        print(f"file save {filename}")
        count += 1

        h, w = frame.shape[:2]
        cropped_frame = frame[35:h-35, 35:w-35]

        padding_px = 50
        blue_color_bgr = (255, 0, 0) 
        
        padded_frame = cv.copyMakeBorder(
            src=cropped_frame,
            top=padding_px,
            bottom=padding_px,
            left=padding_px,
            right=padding_px,
            borderType=cv.BORDER_CONSTANT, 
            value=blue_color_bgr         
        )

        filename = f"image0{count}.jpg"
        while os.path.isfile(filename):
            count+=1
            filename = f"image0{count}.jpg"

        cv.imwrite(filename, padded_frame)
        print(f"file save {filename}")
        count += 1

        cv.imshow("Snapshot preview", padded_frame)
        cv.waitKey(1000)
        cv.destroyWindow("Snapshot preview")
    

    # Press Q on keyboard to exit
    if key & 0xFF == ord('q'):
        break

# Release the video capture
cap.release()

# Close all the frames
cv.destroyAllWindows()
```
